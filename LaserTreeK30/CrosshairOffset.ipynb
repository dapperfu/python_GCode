```python
import gcode

import grbl
```

```python
cnc = grbl.Grbl(port="/dev/ttyUSB0")
```

```python
# Rotated my machine from previous setup.
cnc.direction_port_invert = 0
```

```python
def init(**kwargs):
    """
    Basic Machine Init.
    """
    prog = gcode.GCode(**kwargs)
    prog.G21()
    prog.G90()
    prog.G92(X=0, Y=0)
    prog.G1(F=200)
    prog.G0(F=200)
    return prog
```

```python
prog = init(machine=cnc)

prog.M3(S=10)
prog.G1(X=0, Y=0)
prog.G4(P=2)
prog.M5()
```

```python
prog.run()
```
